```python

print(number)
```

```python
frq = {}
result = []

for i in number:
    if i in frq:
        frq[i] += 1
    else:
        frq[i] = 1

for num in frq:
    for j in range(frq[num]):
        result.append(num)

for i in range(len(result)):
    for j in range(i+1 ,len(result)):
        if frq[result[i]] < frq[result[j]]:
            result[i] , result[j] = result[j] , result[i]

print(result)
```

#### Find non-repeating elements

```python
num = [1,3,4,2,6,3,6,9,4,6,8,9,5]

non_rep = []

for i in num:
    count = 0
    for j in num:
        if i == j:
            count += 1
    if count < 2:
        non_rep.append(i)

print(non_rep)
```

#### Find the second-largest unique number without sort() or max()

```python
num = [1,3,4,2,6,3,6,9,4,6,8,9,5]

largest = -1
sec_largest = -1 

for i in num:
    if i > largest:
        sec_largest = largest
        largest = i
    elif i > sec_largest and i != largest:
        sec_largest = i

if sec_largest == None:
    print("Second Largest Number is not found")
else:
    print("Second Largest :", sec_largest)
```

#### Move all zeros to the end

```python
numbers = [0, 5, 0, 3, 8, 0, 2]
```

```python
zero_count =0
resu = []
for i in numbers:
    if i == 0:
        zero_count += 1
    else:
        resu.append(i)
for i in range(zero_count):
    resu.append(0)
print(resu)
```

#### Group names based on their starting letter

```python
names = ["John", "James", "David", "Daniel", "Alex", "Andrew"]
```

```python


result = {}

for i in names:
    f_name = i[0]

    if f_name in result:
        result[f_name].append(i)
    else:
        result[f_name] = [i]

print(result)
```

#### Find the largest number in each row

```python
matrix = [
    [10, 50, 30],
    [80, 40, 60],
    [20, 90, 70]
]

lar = []

for row in matrix:
    largest = 0
    for i in row:
        if i > largest:
            largest = i

    lar.append(largest)

print(lar)

    
```

#### Find students who passed and calculate their average

```python
students = [
    {"name": "John", "marks": 85},
    {"name": "David", "marks": 35},
    {"name": "Alex", "marks": 65},
    {"name": "Peter", "marks": 40},
    {"name": "Sam", "marks": 90}
]
```

```python
total= 0
name = []

for i in students:

    if i["marks"] > 35:
        name.append(i["name"])
        total += i["marks"]
print("\n============PASSED STUDENTS==================")
for i in name:
    print(i)
print("\n====================AVG MARK=====================")
print(total/len(name))
```

#### Find all duplicate numbers and their frequencies

```python
numbers = [10, 20, 10, 30, 20, 10, 40, 50, 20]

check = []

for i in numbers:
    count = 0
    if i not in check:
        check.append(i)
        for j in numbers:
            if i == j:
                count +=1
        if count >1:
            print(f"number : {i} , appear : {count}")

```

#### Separate even and odd numbers while preserving order

```python
numbers = [12, 5, 8, 7, 3, 10, 15, 20]
odd = []
even = []

for i in numbers:
    if i%2==0:
        even.append(i)
    else:
        odd.append(i)
print("EVEN :", even)
print("ODD", odd)
```
