#### **BANK**

```python
# Account holder

# Balance

# Instance method: deposit

# Instance method: withdraw

# Instance method: display balance

# Class attribute: bank name

# Class method: change bank name

# Static method: validate whether the withdrawal amount is positive.
```

```python
class Bank:

    #class attribute
    bank_name = "ABC Bank"

    #constructor
    def __init__(self,name,balance):
        self.name = name
        self.balance = balance


    #instance method
    def deposite(self,amount):
        if amount > 0:
            self.balance += amount
            print(f"Amount of {amount} is successfully credited")
        else:
            print("Invalid Amount")

    #instance method
    def withdraw(self,amount):
        if not self.validate_amount(amount):
            print("Invalid Amount")
        elif amount > self.balance:
            ("Insufficient balance. ")
        else:
            self.balance -=amount
            print(f"The amount of {amount} is successfully debited")


    def display(self):
        print("-------------Account Details------------\n")
        print("Account Holder:", self.name)
        print("Bank Name:", Bank.bank_name)
        print("Available Balance: ₹",self.balance)


    #class method
    @classmethod
    def change(cls,Bname):
        Bank.bank_name = Bname
        print("Bank name changed successfully")



    @staticmethod
    def validate_amount(amount):
        return amount>0





b1 = Bank("Gifton_Shibu",5000)
b1.deposite(4000)
b1.display()
Bank.change("SBI")
b1.withdraw(500)
b1.display()

    
        
```

#### **STUDENTS MANAGEMENT**

```python
# 1. Beginner – Student Management System
# - Student name
# - Roll number
# - Marks
# - Instance method: calculate average marks
# - Instance method: display student details
# - Class attribute: college name
# - Class method: change college name
# - Static method: validate marks (0–100)
```

```python
class Student:
    school = "Prabhath"
    def __init__(self,name , rollno , *marks):
        self.name = name
        self.rollno = rollno
        self.mark = marks

    def cal_avg(self):
        if len(self.mark) == 0:
            self.avg = 0 
        else:
            self.avg = sum(self.mark) / len(self.mark)
        return self.avg    

    def avg_mark(self):
        if len(self.mark) == 0:
            print("No mark is entered")
        else:
            total = sum(self.mark)
            count = len(self.mark)
            self.avg = total/count
            print(F"Average : {round(self.avg,2)}")

    def display(self):
        print("\n-------------Student Details------------\n")
        print("Student Name:", self.name)
        print("Roll Number:",self.rollno )
        print("Marks: ",self.mark)        


    def update_mark(self, index ,new_mark):
        if not self.T_mark(new_mark):
            print("Enter valid mark")
            return
        if index < 0 or index >= len(self.mark):
            print("Invalid index")
            return
        mark = list(self.mark)
        mark[index] = new_mark
        self.mark = tuple(mark)
        self.cal_avg()
        print("Mark updated successfully")

    def add_mark(self,*mark):
        self.mark = self.mark + mark
        self.cal_avg()
        print("Mark added successfully and ")


    def remove(self,index):
        if index < 0 or index >= len(self.mark):
            print("Invalid index")
            return
        mark = list(self.mark)
        rm = mark.pop(index)
        self.mark = tuple(mark)
        self.cal_avg()
        print(f"Mark removed successfully and the removed is {rm}")

    def grade(self):
        if self.avg >= 90:
            print("S Grade")
        elif self.avg >=70 and self.avg <=89:
            print("First clss")

        elif self.avg >=35  and self.avg < 70:
            print(f"Pass with Average mark {round(self.avg,2)}")

        else:
            print("Failed")


    @classmethod
    def change_school(cls,name):
        cls.school = name
        print("School name is changes successfully")

    @staticmethod
    def T_mark(mark):
        return 0 <= mark <= 100

        
s1 = Student("gifton" , 11, 50,40,20 ,50,60)
s1.display()
s1.avg_mark()
s1.grade()
s1.add_mark(70,90,95)
s1.remove(2)
s1.update_mark(1,60)
s1.display()
s1.grade()
```

#### **Hotel Room Booking System**

```python

# 4. Advanced – Hotel Room Booking System
# - Customer name
# - Room number
# - Number of days
# - Price per day
# - Instance method: book room
# - Instance method: check room availability
# - Instance method: check out
# - Instance method: calculate total bill
# - Instance method: display booking details
# - Class attribute: hotel name
# - Class method: change hotel name
# - Static method: validate booking duration
```

```python
class Hotel:

    Hotel_name = "Novel"
    booked_room = []

    @staticmethod
    def validate_day(day):
        return day>0

    def __init__(self,name,roomno,days,cash):
        self.name = name
        self.rno = roomno
        self.days = days
        self.cash = cash

    def avail_room(self):
        if  self.rno not in Hotel.booked_room:
            print(f"Room number {self.rno} is available")
            return True
        else:
            print(f"Room number {self.rno} is already booked ")
            return False

    def book_room(self):
        if not self.validate_day(self.days):
            print("Invalid Days")
            return
    
        if self.avail_room():
            Hotel.booked_room.append(self.rno)
            print(f"The room number {self.rno} is successfully booked")


    def check_out(self):
        if self.rno in Hotel.booked_room:
            Hotel.booked_room.remove(self.rno)
            print(f"Room number {self.rno} Successfully check-out")
            print("Final Bill : " ,self.bill())
        else:
            print("Invalid Room Number")


    def bill(self):
        self.amount = self.days *self.cash
        return self.amount



    def display_details(self):

        print("\n------ Booking Details ------\n")
        print("Hotel:",Hotel.Hotel_name )
        print("Customer:", self.name)
        print("Room:", self.rno)
        print("Days:",self.days )
        print("Price per day:",self.cash )

        if self.rno in Hotel.booked_room:
            print("Status: Booked")
            print("Total Bill:", self.bill())
        else:
            print("Status: Not Booked")
        print("\n---------------------------------\n")

    @classmethod
    def change_name(cls,name):
        cls.Hotel_name = name
        print("Hotel name changed successfully.")


h1 = Hotel("gifton" ,101,5,2000)
h1.avail_room()
h1.book_room()
h1.display_details()
h2 = Hotel("blesson", 101,3,1500)
h2.avail_room()
h2.book_room()
h1.bill()
h2.check_out()
        
```

#### **Online Payment**

```python
# Parent class Payment with process_payment().

# Child class UPIPayment with its own process_payment().

# Child class CardPayment with its own process_payment().
```

```python
class Payment:

    def __init__(self,amount):
        if amount <+0:
            ValueError("Enter the valid amount")
        self.amount = amount

    def process_payment(self):
        print("Payment is starting process")

        return True

class UPIPayment(Payment):

    def __init__(self,amount,upiid):
        super().__init__(amount)

        self.upiid =upiid

    def process_payment(self):
        print("UPI Payment Started")
        print("UPI ID :",self.upiid)
        print("Amount :" ,self.amount)

        super().process_payment()

        print("The UPI payment successfully completed")

        return True

class CardPayment(Payment):

    def __init__(self,amount,card_number):
        super().__init__(amount)
        self.card_number = card_number

    def process_payment(self):
        print("Card Payment Started")
        print("Card Number :",self.card_number[-4:])
        print("Amount :" ,self.amount)

        super().process_payment()

        print("The Card payment successfully completed")

        return True

upi = UPIPayment(5000 , "gifton@sbi")
card = CardPayment(1000 , "123456789012")

upi.process_payment()

print("---------------------------------\n")

card.process_payment()

print("---------------------------------\n")
```

#### **DRIVING CAR**

```python

# ==================================
# 1. ENGINE CLASS
# ==================================
class Engine:

    def __init__(self):
        self.__is_running = False

    @property
    def running(self):
        return self.__is_running

    def start(self):

        if self.__is_running:
            print("Engine is already started")
            return
        self.__is_running = True
        print("Engine is Turned ON")

    def stop(self):

        if not self.__is_running:
            print("Engine is already turn off")
            return

        self.__is_running = False
        print("Engine is Turned OFF")

# ==================================
# 2. TRANSMISSION CLASS
# ==================================


class Transmission:

    def __init__(self):
        self.__gear = 0

    @property
    def gear(self):
        return self.__gear


    def gear_trasmission(self,gear):
        if gear < 0 or gear > 5:
            print("Wrong Gear")
            return False

    
        self.__gear = gear
    
        if gear == 0:
            print("Gear shifted to Neutral")
        else:
            print(f"Gear chnage to {gear}")
        return True
        


# ==================================
# 3. BRAKE SYSTEM CLASS
# ==================================

class Brake:

    def apply_brake(self,speed):

        if speed <= 0:
            print("All ready stopped")
            return 0

        new_speed = max(0,speed - 20)
        print("Brake Applied")

        return new_speed
        
```

```python
from abc import ABC , abstractmethod

# ==================================
# 4. ABSTRACT VEHICLE CLASS
# ==================================


class Vehicle(ABC):

    @abstractmethod
    def stop(self):
        pass

    @abstractmethod
    def drive(self):
        pass

    @abstractmethod
    def start(self):
        pass

class Car(Vehicle):

    def __init__(self,brand,model_name):
        self.brand = brand
        self.model_name = model_name

        self.engine = Engine()
        self.transmission = Transmission()
        self.brak = Brake()

        self.__speed = 0

    @property
    def speed(self):
        return self.__speed

    def start(self):

        print(f"Starting {self.brand}  {self.model_name}")

        self.engine.start()

    def stop(self):
        self.__speed = 0

        self.transmission.gear_trasmission(0)

        self.engine.stop()

        print("Car stopped completely")


    def drive(self):

        if  not self.engine.running:
            print("Turn ON The Engine")
            return

        if self.transmission.gear == 0:
            print("Change The Gear To Drive")
            return

        self.__speed += 20

        print(f"Car is moving at {self.speed} km/h")

    def accelerate(self):

        if  not self.engine.running:
            print("Turn ON The Engine")
            return

        if self.transmission.gear == 0:
            print("Change The Gear To Drive")
            return
        
        self.__speed += 10

        print(f"Car is moving at {self.speed} km/h")

    def brake(self):
        self.__speed = self.brak.apply_brake(self.__speed)

        print(f"Current speed: {self.__speed} km/h")




        
```

```python
#==================================
# 6. OBJECT CREATION
# ==================================

car = Car("Toyota", "Innova")


# ==================================
# 7. DRIVING SIMULATION
# ==================================

car.start()
car.transmission.gear_trasmission(1)

car.drive()

car.accelerate()

car.transmission.gear_trasmission(2)

car.accelerate()

car.brake()

car.brake()

car.stop()
```
