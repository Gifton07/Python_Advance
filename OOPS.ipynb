### Static method vs Class method vs Instance methods

#### **BANK**

```python
# Account holder

# Balance

# Instance method: deposit

# Instance method: withdraw

# Instance method: display balance

# Class attribute: bank name

# Class method: change bank name

# Static method: validate whether the withdrawal amount is positive.
```

```python
class Bank:

    #class attribute
    bank_name = "ABC Bank"

    #constructor
    def __init__(self,name,balance):
        self.name = name
        self.balance = balance


    #instance method
    def deposite(self,amount):
        if amount > 0:
            self.balance += amount
            print(f"Amount of {amount} is successfully credited")
        else:
            print("Invalid Amount")

    #instance method
    def withdraw(self,amount):
        if not self.validate_amount(amount):
            print("Invalid Amount")
        elif amount > self.balance:
            ("Insufficient balance. ")
        else:
            self.balance -=amount
            print(f"The amount of {amount} is successfully debited")


    def display(self):
        print("-------------Account Details------------\n")
        print("Account Holder:", self.name)
        print("Bank Name:", Bank.bank_name)
        print("Available Balance: ₹",self.balance)


    #class method
    @classmethod
    def change(cls,Bname):
        Bank.bank_name = Bname
        print("Bank name changed successfully")



    @staticmethod
    def validate_amount(amount):
        return amount>0





b1 = Bank("Gifton_Shibu",5000)
b1.deposite(4000)
b1.display()
Bank.change("SBI")
b1.withdraw(500)
b1.display()

    
        
```

#### **STUDENTS MANAGEMENT**

```python
# 1. Beginner – Student Management System
# - Student name
# - Roll number
# - Marks
# - Instance method: calculate average marks
# - Instance method: display student details
# - Class attribute: college name
# - Class method: change college name
# - Static method: validate marks (0–100)
```

```python
class Student:
    school = "Prabhath"
    def __init__(self,name , rollno , *marks):
        self.name = name
        self.rollno = rollno
        self.mark = marks

    def cal_avg(self):
        if len(self.mark) == 0:
            self.avg = 0 
        else:
            self.avg = sum(self.mark) / len(self.mark)
        return self.avg    

    def avg_mark(self):
        if len(self.mark) == 0:
            print("No mark is entered")
        else:
            total = sum(self.mark)
            count = len(self.mark)
            self.avg = total/count
            print(F"Average : {round(self.avg,2)}")

    def display(self):
        print("\n-------------Student Details------------\n")
        print("Student Name:", self.name)
        print("Roll Number:",self.rollno )
        print("Marks: ",self.mark)        


    def update_mark(self, index ,new_mark):
        if not self.T_mark(new_mark):
            print("Enter valid mark")
            return
        if index < 0 or index >= len(self.mark):
            print("Invalid index")
            return
        mark = list(self.mark)
        mark[index] = new_mark
        self.mark = tuple(mark)
        self.cal_avg()
        print("Mark updated successfully")

    def add_mark(self,*mark):
        self.mark = self.mark + mark
        self.cal_avg()
        print("Mark added successfully and ")


    def remove(self,index):
        if index < 0 or index >= len(self.mark):
            print("Invalid index")
            return
        mark = list(self.mark)
        rm = mark.pop(index)
        self.mark = tuple(mark)
        self.cal_avg()
        print(f"Mark removed successfully and the removed is {rm}")

    def grade(self):
        if self.avg >= 90:
            print("S Grade")
        elif self.avg >=70 and self.avg <=89:
            print("First clss")

        elif self.avg >=35  and self.avg < 70:
            print(f"Pass with Average mark {round(self.avg,2)}")

        else:
            print("Failed")


    @classmethod
    def change_school(cls,name):
        cls.school = name
        print("School name is changes successfully")

    @staticmethod
    def T_mark(mark):
        return 0 <= mark <= 100

        
s1 = Student("gifton" , 11, 50,40,20 ,50,60)
s1.display()
s1.avg_mark()
s1.grade()
s1.add_mark(70,90,95)
s1.remove(2)
s1.update_mark(1,60)
s1.display()
s1.grade()
```

#### **Hotel Room Booking System**

```python

# 4. Advanced – Hotel Room Booking System
# - Customer name
# - Room number
# - Number of days
# - Price per day
# - Instance method: book room
# - Instance method: check room availability
# - Instance method: check out
# - Instance method: calculate total bill
# - Instance method: display booking details
# - Class attribute: hotel name
# - Class method: change hotel name
# - Static method: validate booking duration
```

```python
class Hotel:

    Hotel_name = "Novel"
    booked_room = []

    @staticmethod
    def validate_day(day):
        return day>0

    def __init__(self,name,roomno,days,cash):
        self.name = name
        self.rno = roomno
        self.days = days
        self.cash = cash

    def avail_room(self):
        if  self.rno not in Hotel.booked_room:
            print(f"Room number {self.rno} is available")
            return True
        else:
            print(f"Room number {self.rno} is already booked ")
            return False

    def book_room(self):
        if not self.validate_day(self.days):
            print("Invalid Days")
            return
    
        if self.avail_room():
            Hotel.booked_room.append(self.rno)
            print(f"The room number {self.rno} is successfully booked")


    def check_out(self):
        if self.rno in Hotel.booked_room:
            Hotel.booked_room.remove(self.rno)
            print(f"Room number {self.rno} Successfully check-out")
            print("Final Bill : " ,self.bill())
        else:
            print("Invalid Room Number")


    def bill(self):
        self.amount = self.days *self.cash
        return self.amount



    def display_details(self):

        print("\n------ Booking Details ------\n")
        print("Hotel:",Hotel.Hotel_name )
        print("Customer:", self.name)
        print("Room:", self.rno)
        print("Days:",self.days )
        print("Price per day:",self.cash )

        if self.rno in Hotel.booked_room:
            print("Status: Booked")
            print("Total Bill:", self.bill())
        else:
            print("Status: Not Booked")
        print("\n---------------------------------\n")

    @classmethod
    def change_name(cls,name):
        cls.Hotel_name = name
        print("Hotel name changed successfully.")


h1 = Hotel("gifton" ,101,5,2000)
h1.avail_room()
h1.book_room()
h1.display_details()
h2 = Hotel("blesson", 101,3,1500)
h2.avail_room()
h2.book_room()
h1.bill()
h2.check_out()
        
```
