### Student Data Management 


```python
students = {
    "S101": {
        "name": "Arun",
        "marks": 85
    },
    "S102": {
        "name": "Rahul",
        "marks": 90
    }
}
```

#### Add a student

```python
def add_student(std_id,name, marks):
    students[std_id] = {
        "name": name,
        "marks": marks
    }
```

#### Display all students

```python
def all_students():
    for std_id,std_info in students.items():
        print(f'ID:{std_id} , Name: {std_info["name"]} , Marks: {std_info["marks"]}\n')
```

#### Search using student ID

```python
def student_info(std_id):
    value = students.get(std_id)
    if value:
        print(f'Name :{value["name"]}, Mark: {value["marks"]}\n')
    else:
        print("This ID student is not in the DB\n")


```

#### Update marks

```python
def update_mark(std_id,new_mark):
    value = students.get(std_id)
    if value:
        value['marks'] =new_mark
        print("Mark updated successfully\n")
    else:
        print("This ID student is not in the DB\n")        
```

#### Calculate average marks

```python
def avg_mark():
    total = sum((i["marks"] for i in students.values()))
    avg_mark = total/len(students)
    print(f'Average mark = {avg_mark}\n')
```
