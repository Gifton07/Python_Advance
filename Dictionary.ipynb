### **Student Data Management**


```python
students = {
    "S101": {
        "name": "Arun",
        "marks": 85
    },
    "S102": {
        "name": "Rahul",
        "marks": 90
    }
}
```

#### Add a student

```python
def add_student(std_id,name, marks):
    students[std_id] = {
        "name": name,
        "marks": marks
    }
```

#### Display all students

```python
def all_students():
    for std_id,std_info in students.items():
        print(f'ID:{std_id} , Name: {std_info["name"]} , Marks: {std_info["marks"]}\n')
```

#### Search using student ID

```python
def student_info(std_id):
    value = students.get(std_id)
    if value:
        print(f'Name :{value["name"]}, Mark: {value["marks"]}\n')
    else:
        print("This ID student is not in the DB\n")


```

#### Update marks

```python
def update_mark(std_id,new_mark):
    value = students.get(std_id)
    if value:
        value['marks'] =new_mark
        print("Mark updated successfully\n")
    else:
        print("This ID student is not in the DB\n")        
```

#### Calculate average marks

```python
def avg_mark():
    total = sum((i["marks"] for i in students.values()))
    avg_mark = total/len(students)
    print(f'Average mark = {avg_mark}\n')
```

## **Library Book Management System**

```python
library = [

    {
        "book_id" :101,
        "title" : "Python Crash Course",
        "author" : "Eric Matthes",
        "is_available": True
    },

    {
        "book_id": 102,
        "title": "Atomic Habits",
        "author": "James Clear",
        "is_available": True
    },
    
    {
        "book_id": 103,
        "title": "Clean Code",
        "author": "Robert C. Martin",
        "is_available": False
    }
]
```

#### Add a Book

```python
def add_book(id , name , au_name , avail = True ):
    new_book = {
        "book_id": id ,
        "title" : name,
        "author" : au_name,
        "is_available" : avail
    }

    library.append(new_book)
    print("Successfully book is added\n")
```

#### Display All Books:


```python
def display_book():
    print("\n*************************************************\n")
    print("                  ALL THE BOOKS")
    print("\n*************************************************\n")
    for i in library:
        # print(f'Book_ID : {library[i]["book_id"]}')
        print("\nBook ID:", i["book_id"])
        print("Book Name: ", i["title"])
        print("Author Name: ", i["author"])

        if i["is_available"] == True:
            print("Book is available\n")
        else:
            print("Book is borrowed\n")

    print("\n*************************************************\n")         
            
# display_book()
```

#### Search a Book

```python
def search_book(search):
    found = False

    for book in library:
        if search.lower() in book["title"].lower()  or  search.lower() in book["author"].lower():
            print("\n*************************************************")
            print("                  ITEM IS FOUND")
            print("*************************************************\n")
            print("\nBook ID:", book["book_id"])
            print("Book Name: ", book["title"])
            print("Author Name: ", book["author"])

            if book["is_available"] == True:
                print("Book is available\n")
            else:
                print("Book is borrowed\n")

            found = True

    if found == False:
        print("\n*************************************************")
        print("                  ITEM IS NOT FOUND")
        print("*************************************************\n")       

# search_book("Python")     
```

#### Borrow a Book

```python
def borrow(id):
    for book in library:
        if id == book["book_id"]:
            if book["is_available"] == True:
                book["is_available"] = False
                print("\n*************************************************************")
                print(f"        SUCCESSFULLY UPDATE THE BORROW  BOOK_ID : {id}")
                print("**************************************************************\n")  
            else:
                print("Sorry! This book is already borrowed.")
            return
    
    print("The book not found")

# borrow(103)
```

#### Return a Book

```python
def return_book(id):
    for book in library:
        if book["book_id"] == id:
            if book["is_available"] == False:
                book["is_available"] = True
                print("\n*************************************************")
                print(f'   SUCCESSFULLY UPDATE THE RETURN BOOK_ID : {id}')
                print("*************************************************\n")  
            else:
                print("Sorry! This book is already return.")
            return   

    print("Book not found")

# return_book(102)          
```

#### Display Available Books

```python
def avail_book():
    print("\n*************************************************")
    for book in library:
        if book["is_available"] == True:
            print("\nBook ID:", book["book_id"])
            print("Book Name: ", book["title"])
            print("Author Name: ", book["author"],"\n")  
    print("\n*************************************************")
# avail_book()          
```

#### Delete the book

```python
def del_book(id):
    for book in library:
        if book["book_id"] == id:
            library.remove(book)
            print("\n*************************************************")
            print("          SUCCESSFULLY REMOVE FROM THE DATABASE")
            print("*************************************************\n") 
            return

    print("Book not in database")

# del_book(101)
            
```

```python
print("\n--- Library Management Menu---")
print("1. Add Book")
print("2. Display all Books")
print("3. Search a Book using BOOK NAME/ AUTHOR NAME")
print("4. Borrow a book")
print("5. Return a book")
print("6. Display available  books")
print("7. Delete a book from DB")
print("8. Exit\n")
```

```python
while True:
    choice = int(input("Enter your choice"))

    match choice:
        case 1 :
            id = int(input("Enter BOOK_ID"))
            name = input("Enter Book Name")
            au_name = input("Enter Author Name")
            add_book(id,name,au_name)

        case 2 :
            display_book()

        case 3:
            name = input("Enter BOOK NAME/ AUTHOR NAME")
            search_book(name)
        
        case 4:
            id = int(input("Enter BOOK_ID"))
            borrow(id)

        case 5:
            id = int(input("Enter BOOK_ID"))
            return_book(id)

        case 6:
            avail_book()

        case 7:
            id = int(input("Enter BOOK_ID"))
            del_book(id)

        case 8:
            print("\n*************************************************")
            print("Closing the program")
            print("\n*************************************************")
            break

        case _:
            print("Invalid choice. Please enter a number from 1 to 8.")
```
