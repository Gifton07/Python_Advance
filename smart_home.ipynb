### **Project: Smart Home Automation System**

#### **1. Objective**
Build a Python-based Smart Home Automation System using OOP to manage smart lights, air conditioners, and smart doors. Focus on backend logic without GUI or database.

#### **2. Class Requirements**

##### **1. SmartDevice (Abstract Class)**
- Attributes: `device_name`, `__is_on`
- Abstract methods: `turn_on()`, `turn_off()`
- Methods: `device_info()`
- Property: `is_on`

##### **2. SmartLight (Child Class)**
- Attribute: `__brightness` (default: 50)
- Methods: `turn_on()`, `turn_off()`, `set_brightness(value)`
- Property: `brightness`
- Validation: Brightness between 0–100.

##### **3. AirConditioner (Child Class)**
- Attribute: `__temperature` (default: 24°C)
- Methods: `turn_on()`, `turn_off()`, `set_temperature(value)`
- Property: `temperature`
- Validation: Temperature between 16–30°C.

##### **4. SmartDoor (Child Class)**
- Attribute: `__is_locked` (default: True)
- Methods: `turn_on()`, `turn_off()`, `lock()`, `unlock()`
- Property: `is_locked`

##### **5. SmartHome (Controller Class)**
- Attributes: `home_name`, `devices`
- Methods:
  - `add_device(device)`
  - `show_devices()`
  - `turn_on_all()`
  - `turn_off_all()`
- Prevent duplicate devices.

#### **3. OOP Concepts**
- Abstraction
- Inheritance
- Encapsulation
- Polymorphism
- Method Overriding
- Composition
- Constructors
- Properties
- Input Validation
- State Management

#### **4. Expected Workflow**
1. Create Light, AC, and Door objects.
2. Create a SmartHome object.
3. Add all devices to the home.
4. Display device information.
5. Turn on the light and set brightness to 80%.
6. Turn on the AC and set temperature to 22°C.
7. Activate and unlock the smart door.
8. Display updated device states.
9. Turn off all devices.
10. Display final states.

#### **5. Validation**
- Prevent repeated ON/OFF operations.
- Reject invalid brightness and temperature.
- Prevent repeated door lock/unlock operations.
- Prevent duplicate device registration.
- Ensure invalid operations do not change device states.

#### **6. Implementation Rules**
- Implement one class at a time.
- Test each class independently.
- Use proper encapsulation and validation.
- Build the SmartHome controller last.
- No GUI, database, or external APIs.
