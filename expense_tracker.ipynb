#### Add Expense

```python
def add_expanse(title,amount,category,date):
    if tracker:
        exp_id = max(id["exp_id"] for id in tracker)+1
    else:
        exp_id = 1
    
    new_exp = {
        "exp_id" :exp_id,
        "title" : title,
        "amount" : amount,
        "category" : category,
        "date" : date
    }

    tracker.append(new_exp)

    print(f"New expanse of {amount} ruppes added successfuly")

add_expanse("sancks",50,"Food","2026-11-18")
```

#### Display Expenses

```python
def display_expense():
    
    if not tracker:
        print("No Expense")
        return
    
    print("================ EXPENSE DETAILS ===================")

    for exp in tracker:
        print("\nExpanse ID : ", exp["exp_id"])
        print("Title : ", exp["title"])
        print("Amount ₹  : ", exp["amount"])
        print("Category : ", exp["category"])
        print("Date  : ", exp["date"],"\n")

    print("====================================================")

# display_expense()
```

#### Search Expense

```python
def search_exp(search):

    found = False
    

    
    print("\n--------------------------")    
    
    for exp in tracker:

        if search.lower() == exp["title"].lower() or search.lower() == exp["category"].lower() :
            

            print("\nExpense ID:", exp["exp_id"])
            print("Title:", exp["title"])
            print("Amount: ₹", exp["amount"])
            print("Category:", exp["category"])
            print("Date:", exp["date"],"\n")

            found = True
        
        
    if found == False:
        print("The item is not available")
    
    print("\n--------------------------")

# search_exp("food")
```

#### Delete Expense

```python
def del_exp(id):

    for exp in tracker:

        if id == exp["exp_id"]:
            tracker.remove(exp)
            print("\n------------------------------------------------------")
            print(f" Successfully Delete the ID : {id} having amount of {exp["amount"]}")
            print("-------------------------------------------------------\n")

            return
        
    print("ID is not found")

# del_exp(1)
```

#### Total Expenses

```python
def total():

    print("\n===============TOTAL EXPANSE===================\n")
    total = 0
    for exp in tracker:
        total += exp["amount"]
    print("           Total Expenses: ₹", total)
    print("\n===============================================\n")

# total()
```
